```python
import os
import shutil

# Create destination directory if it doesn't exist
dest_dir = "/kaggle/working/BinaryDetectionWeights"
os.makedirs(dest_dir, exist_ok=True)

model_paths = [
    # Yolov8 weights
    '/kaggle/input/rasoul-b-e-s-obb-yolov8n/yolo_training_results/yolov8n-obb/weights/best.pt',
    '/kaggle/input/rasoul-b-e-s-obb-yolov8s/yolo_training_results/yolov8s-obb/weights/best.pt',
    '/kaggle/input/rasoul-b-e-s-obb-yolov8m/yolo_training_results/yolov8m-obb/weights/best.pt',

    # Yolo11 weights
    '/kaggle/input/rasoul-b-e-s-obb-yolo11n-s/yolo_training_results/yolo11n-obb/weights/best.pt',
    '/kaggle/input/rasoul-b-e-s-obb-yolo11n-s/yolo_training_results/yolo11s-obb/weights/best.pt',
    '/kaggle/input/rasoul-b-e-s-obb-yolo11m/yolo_training_results/yolo11m-obb/weights/best.pt'
]

for path in model_paths:
    model_name = os.path.basename(os.path.dirname(os.path.dirname(path)))  # e.g., yolov8n
    dest_path = os.path.join(dest_dir, f"{model_name}.pt")
    shutil.copy(path, dest_path)
    print(f"Copied {path} to {dest_path}")
```
