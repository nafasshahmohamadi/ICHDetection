```python
import os
import shutil

# Create destination directory if it doesn't exist
dest_dir = "/kaggle/working/MulticlassDetectionWeights"
os.makedirs(dest_dir, exist_ok=True)

model_paths = [
    # Yolov8 weights
    '/kaggle/input/rasoul-multiclass-detection-yolov8n-s/yolo_training_results/yolov8n/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolov8n-s/yolo_training_results/yolov8s/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolov8m/yolo_training_results/yolov8m/weights/best.pt',

    # Yolo11 weights
    '/kaggle/input/rasoul-multiclass-detection-yolo11n-s/yolo_training_results/yolo11n/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolo11n-s/yolo_training_results/yolo11s/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolo11m/yolo_training_results/yolo11m/weights/best.pt',

    # Yolo12 weights
    '/kaggle/input/rasoul-multiclass-detection-yolo12n-s/yolo_training_results/yolo12n/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolo12n-s/yolo_training_results/yolo12s/weights/best.pt',
    '/kaggle/input/rasoul-multiclass-detection-yolo12m/yolo_training_results/yolo12m/weights/best.pt',
]

for path in model_paths:
    model_name = os.path.basename(os.path.dirname(os.path.dirname(path)))  # e.g., yolov8n
    dest_path = os.path.join(dest_dir, f"{model_name}.pt")
    shutil.copy(path, dest_path)
    print(f"Copied {path} to {dest_path}")
```

```python
import os
destination_dir = "/kaggle/input/rasoul-multiclass-detection-yolo12n-s/dataset/images/val"

def count_files_in_directory(directory):
    total_files = 0
    # Iterate through each item in the directory
    for item in os.listdir(directory):
        item_path = os.path.join(directory, item)
        # If the item is a directory, recursively count files in it
        if os.path.isdir(item_path):
            num_files_in_subfolder = count_files_in_directory(item_path)
            print(f"Folder '{item}': {num_files_in_subfolder} files")
            total_files += num_files_in_subfolder
        # If the item is a file, increment the file count
        elif os.path.isfile(item_path):
            total_files += 1
    return total_files

# Check if the destination directory exists
if os.path.exists(destination_dir):
    total_files = count_files_in_directory(destination_dir)
    print(f"Total files in '{destination_dir}': {total_files}")
else:
    print(f"The directory '{destination_dir}' does not exist.")
```
