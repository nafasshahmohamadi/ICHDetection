```python
from ultralytics import YOLO
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
os.environ["WANDB_MODE"] = "offline"
```

```python
path = "/kaggle/working/organized-bc-dataset"
```

```python
# Train the model (using default project structure in 'runs/')
model = YOLO('yolov8s-cls.pt')  # Or use your own model if applicable
results = model.train(
    data=path,                          # Path to the dataset
    epochs=50,                          # Number of epochs
    batch=32,                           # Batch size
    imgsz=640,                          # Image size
    project='y8s-bc',                     # Default project folder (no full paths)
    name='testing'                      # Name of the experiment folder
)
```

```python
# Load YOLO training results
csv_path = '/kaggle/working/y8s-bc/testing/results.csv'  # Adjust path if needed
df = pd.read_csv(csv_path)

# Display all columns for reference
print("Available columns:", df.columns)
```

```python
# Plot Training and Validation Loss
plt.figure(figsize=(10, 6))
plt.plot(df['epoch'], df['train/loss'], label='Training Loss')
plt.plot(df['epoch'], df['val/loss'], label='Validation Loss')

plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss Curves')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Evaluate on validation set
val_results = model.val(data=path)

# Evaluate on test set
test_results = model.val(data=path)
```

```python
# Plot Accuracy Metrics
plt.figure(figsize=(10, 6))
plt.plot(df['epoch'], df['metrics/accuracy_top1'], label='Top-1 Accuracy')
plt.plot(df['epoch'], df['metrics/accuracy_top5'], label='Top-5 Accuracy')

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Top-1 and Top-5 Accuracy Curves')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def show_image(path, title):
    img = mpimg.imread(path)
    plt.figure(figsize=(25, 25))
    plt.imshow(img)
    plt.title(title, fontsize=16)
    plt.axis('off')
    plt.grid(False)
    plt.show()

# Show confusion matrices with high resolution
show_image('/kaggle/working/y8s-bc/testing/confusion_matrix.png', 'Confusion Matrix (Raw)')
show_image('/kaggle/working/y8s-bc/testing/confusion_matrix_normalized.png', 'Confusion Matrix (Normalized)')
```

```python
# Show validation batch labels and predictions
for i in range(3):
    plt.figure(figsize=(25, 25))

    plt.subplot(1, 2, 1)
    label_img = mpimg.imread(f'/kaggle/working/y8s-bc/testing/val_batch{i}_labels.jpg')
    plt.imshow(label_img)
    plt.title(f'Validation Batch {i} - Labels')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    pred_img = mpimg.imread(f'/kaggle/working/y8s-bc/testing/val_batch{i}_pred.jpg')
    plt.imshow(pred_img)
    plt.title(f'Validation Batch {i} - Predictions')
    plt.axis('off')

    plt.show()
```
